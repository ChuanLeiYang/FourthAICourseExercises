# 1. <a id='toc1_'></a>[Open Compass框架介绍](#toc0_)

## 1.1. <a id='toc1_1_'></a>[什么是OpenCompass?](#toc0_)

 OpenCompass 是一个开源项目，旨在为机器学习和自然语言处理领域提供多功能、易于使用的工具和框架。其中包含的多个开源模型和开源数据集（BenchMarks），方便进行模型的效果评测。


# 2. <a id='toc2_'></a>[生成式大模型的评估指标及OpenCompass支持情况](#toc0_)

## 2.1. <a id='toc2_1_'></a>[核心评估指标](#toc0_)

OpenCompass支持以下主要评估指标，覆盖生成式大模型的多样化需求：

准确率（Accuracy）：用于选择题或分类任务，通过比对生成结果与标准答案计算正确率。在OpenCompass中通过metric=accuracy配置

困惑度（Perplexity, PPL）：衡量模型对候选答案的预测能力，适用于选择题评估。需使用ppl类型的数据集配置（如ceval_ppl）

生成质量（GEN）：通过文本生成结果提取答案，需结合后处理脚本解析输出。使用gen类型的数据集（如ceval_gen），配置metric=gen并指定
后处理规则

ROUGE/LCS：用于文本生成任务的相似度评估，需安装rouge==1.0.1依赖，并在数据配置中设置metric=rouge

条件对数概率（CLP）：结合上下文计算答案的条件概率，适用于复杂推理任务，需在模型配置中启用use_logprob=True


# 3. <a id='toc3_'></a>[支持的开源评估数据集及使用差异](#toc0_)

## 3.1. <a id='toc3_1_'></a>[主流开源数据集](#toc0_)

OpenCompass内置超过70个数据集，覆盖五大能力维度：

知识类：C-Eval（中文考试题）、CMMLU（多语言知识问答）、MMLU（英文多选题）。

推理类：GSM8K（数学推理）、BBH（复杂推理链）。

语言类：CLUE（中文理解）、AFQMC（语义相似度）。

代码类：HumanEval（代码生成）、MBPP（编程问题）。

多模态类：MMBench（图像理解）、SEED-Bench（多模态问答）。


## 3.2. <a id='toc3_2_'></a>[数据集区别与选择](#toc0_)

评估范式差异：

_gen后缀数据集：生成式评估，需后处理提取答案（如ceval_gen）

_ppl后缀数据集：困惑度评估，直接比对选项概率（如ceval_ppl）

领域覆盖：

C-Eval：侧重中文STEM和社会科学知识，包含1.3万道选择题

LawBench：法律领域专项评估，需额外克隆仓库并配置路径

## 4.3. <a id='toc4_3_'></a>[配置评估任务](#toc0_)

[配置评估任务](https://opencompass.readthedocs.io/zh-cn/latest/get_started/quick_start.html#id3)

用开源数据集评估的意义：我们微调模型后，如果回答训练的问题很准确，常规问题回答偏差比较大，那有可能过拟合了。所以需要用开源数据集评估一下。

在 OpenCompass 中，每个评估任务由待评估的模型和数据集组成。评估的入口点是 run.py。用户可以通过命令行或配置文件选择要测试的模型和数据集。

对于对话模型

### 4.3.1. <a id='toc4_3_1_'></a>[评估方式一：自定义HF模型](#toc0_)

<img src="./Image/2025-05-12-22-17-38.png" style="margin-left: 0" width="70%">

python run.py \

    --datasets demo_gsm8k_chat_gen demo_math_chat_gen \

    --hf-type chat \

    --hf-path /root/AI-WSL/models/Qwen/Qwen1.5-0.5B-Chat \

    --debug

    其中，
    --datasets：数据集路径  
    --hf-type：模型类型，可选参数，一般模型带后缀Chat，那type就是chat，如果没有，一般可以写base ，这个参数也可以不给，模型会根据配置文件判断是哪个类型。 但是如果给了chat类型，那么本次评估的模型就都要是chat类型的。
    --hf-path：模型路径  
    --debug：输出评估日志

请注意，通过这种方式（自定义HF模型），OpenCompass 一次只评估一个模型，而其他方式可以一次评估多个模型。评估过程是要消耗显存资源的可以注意一下资源消耗

## 4.5. <a id='toc4_5_'></a>[自定义评估](#toc0_)

评估分为两个阶段，第一个阶段是选基座模型的时候，判断模型是否符合我的需求，先用和目标类似的开源数据集来评估一下。比如，偏向数学计算的模型，选数学相关的数据集来评估，看得分判断模型性能。

第二个阶段是微调训练完成后，评估模型训练效果。需要用自定义的数据集来评估。

[自定义数据集](https://opencompass.readthedocs.io/zh-cn/latest/advanced_guides/custom_dataset.html)

根据官网要求自定义数据集，例如。jsonl数据集

![](Image/2025-05-14-00-39-25.png)

自己新建一个.jsonl文件，定义数据集

![](Image/2025-05-14-00-42-38.png)

命令行列表

自定义数据集可直接通过命令行来调用开始评测。官网案例命令如下

python run.py \  
    --models hf_llama2_7b \  
    --custom-dataset-path xxx/test_mcq.csv \  
    --custom-dataset-data-type mcq \  
    --custom-dataset-infer-method ppl  

其实我们只需要模型路径models和数据集路径custom-dataset-path这两个就够了，不过注意如果是本地的模型，要把--models换成--hf-path


python run.py \  
    --hf-path /root/AI-WSL/models/Qwen/Qwen1.5-0.5B-Chat \  
    --custom-dataset-path /root/AI-WSL/data/dataopencompass/test_qa.jsonl \
