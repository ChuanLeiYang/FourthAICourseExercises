## 1.1. <a id='toc1_1_'></a>[WSLInstallation](#toc0_)

参考连接：[WSL安装至 D方法](https://blog.csdn.net/Natsuago/article/details/145594631?spm=1001.2101.3001.6650.14&utm_medium=distribute.pc_relevant.none-task-blog-2%7Edefault%7EBlogCommendFromBaidu%7ERate-14-145594631-blog-124568920.235%5Ev43%5Epc_blog_bottom_relevance_base1&depth_1-utm_source=distribute.pc_relevant.none-task-blog-2%7Edefault%7EBlogCommendFromBaidu%7ERate-14-145594631-blog-124568920.235%5Ev43%5Epc_blog_bottom_relevance_base1&utm_relevant_index=21)

D盘创建 WSL 文件夹，并在该文件夹下创建 Ubuntu-24.04 文件夹。

<img src="./Image/2025-04-29-23-16-12.png" style="margin-left: 0" width="80%">


下载并导出 Ubuntu-24.04

首先，查看可用的 WSL 发行版：

wsl --list --online

如图所示为终端输出：

<img src="./Image/2025-04-29-23-20-43.png" style="margin-left: 0" width="50%">


安装 Ubuntu-24.04

wsl --install -d Ubuntu-24.04

安装完毕后会要求你创建一个新用户，按照提示输入用户名和密码即可，然后按 Ctrl + D 退出即可。

导出 Ubuntu-24.04 为 .tar 文件

wsl --export Ubuntu-20.04 D:\SoftwareInstallationDirectory\Ubuntu-24.04\Ubuntu-24.04.tar

这会下载 Ubuntu-24.04 并将其导出到 D:\SoftwareInstallationDirectory\Ubuntu-24.04\Ubuntu-24.04.tar，然后可以执行ls D:\SoftwareInstallationDirectory\Ubuntu-24.04查看Ubuntu-24.04.tar已经成功导出。

取消注册原有的 Ubuntu-24.04，如果你已经安装了 Ubuntu-24.04（默认在 C 盘），可以将其从 WSL 注销：

wsl --unregister Ubuntu-24.04

导入 Ubuntu-24.04 到 D 盘

运行以下命令，将 Ubuntu-24.04 重新导入到 D:\SoftwareInstallationDirectory\Ubuntu-24.04：

wsl --import Ubuntu-24.04 D:\SoftwareInstallationDirectory\Ubuntu-24.04 D:\SoftwareInstallationDirectory\Ubuntu-24.04\Ubuntu-24.04.tar --version 2

这将会把 Ubuntu-20.04 安装到 D 盘，而不是默认的 C 盘，如图所示正在安装。

在 D:\SoftwareInstallationDirectory\Ubuntu-24.04 目录下，WSL2 发行版的文件存储在一个 虚拟磁盘映像文件（ext4.vhdx） 中，该文件用于存储整个 Ubuntu-24.04 文件系统，如下图所示：

启动 Ubuntu-24.04

导入完成后，你可以启动 WSL：

wsl -d Ubuntu-24.04

## 3.6. <a id='toc3_6_'></a>[制作Dockerfile](#toc0_)

![](Image/2025-04-30-00-11-10.png)

Dockerfile源码如下：

FROM nvcr.io/nvidia/tritonserver:25.03-vllm-python-py3
WORKDIR /app

EXPOSE 8000
CMD ["python3", "-m", "vllm.entrypoints.openai.api_server", \
    "--model", "/app/models/Qwen/Qwen2.5-1.5B-Instruct-merged", \
    "--port", "8000", "--max_model_len", "512", \
    "--tensor-parallel-size", "1", \
    "--gpu_memory_utilization", "0.85", \
    "--max_num_seqs", "128", \
    "--enforce-eager"]

其中，"/app/models/Qwen/Qwen2.5-1.5B-Instruct-merged"指的是模型在容器中的路径，在启动容器的时候要将本地的模型路径映射到这个路径

## 3.7. <a id='toc3_7_'></a>[构建Docker镜像](#toc0_)

docker build -t vllm-image .

## 3.8. <a id='toc3_8_'></a>[启动容器](#toc0_)

docker run -d \
  --name vllm-server \
  -p 8000:8000 \
  -v /root/AI-WSL/models/Qwen/Qwen2.5-1.5B-Instruct-merged:/app/models/Qwen/Qwen2.5-1.5B-Instruct-merged \
  --gpus all \
--shm-size 16G \
  vllm-image


## 3.9. <a id='toc3_9_'></a>[检查容器运行状态](#toc0_)

docker ps -a | grep "vllm-server"

正常状态‌：STATUS 显示为 Up，且端口映射正确
异常处理‌：若状态为 Exited，使用 docker logs vllm-server 查看错误日志（如模型路径错误、显存不足等）


## 3.10. <a id='toc3_10_'></a>[运行多轮对话](#toc0_)

![](Image/2025-04-30-00-24-25.png)

先进入.py文件所在目录

cd /root/AI-WSL/project/dockervllm

给.py文件配置解释器，Ctrl+shift+p

<img src="./Image/2025-04-30-00-26-48.png" style="margin-left: 0" width="80%">


选择一个可以运行.py文件的解释器，这里我选择自己以前创建的一个虚拟环境

<img src="./Image/2025-04-30-00-29-10.png" style="margin-left: 0" width="60%">


test03.py的内容如下：

```python
#多轮对话
from openai import OpenAI

#定义多轮对话方法
def run_chat_session():
    #初始化客户端，api_key是公开的，可以随便写，但是要写
    client = OpenAI(base_url="http://localhost:8000/v1",api_key="suibianxie")
    #初始化对话历史
    chat_history = []
    #启动对话循环
    while True:
        #获取用户输入
        user_input = input("用户：")
        if user_input.lower() == "exit":
            print("退出对话。")
            break
        #更新对话历史(添加用户输入)
        chat_history.append({"role":"user","content":user_input})
        #调用模型回答
        try:
            chat_complition = client.chat.completions.create(messages=chat_history,model="/app/models/Qwen/Qwen2.5-1.5B-Instruct-merged")
            #获取最新回答
            model_response = chat_complition.choices[0]
            print("AI:",model_response.message.content)
            #更新对话历史（添加AI模型的回复）
            chat_history.append({"role":"assistant","content":model_response.message.content})
        except Exception as e:
            print("发生错误：",e)
            break
if __name__ == '__main__':
    run_chat_session()
```

在vllm服务启动的情况下执行.py脚本

python test03.py
