# 1. <a id='toc1_'></a>[Embedding_model](#toc0_)

## 1.1. <a id='toc1_1_'></a>[概念与核心原理](#toc0_)

### 1.1.1. <a id='toc1_1_1_'></a>[嵌入模型的本质](#toc0_)

嵌入模型（Embedding Model）是一种将离散数据（如文本、图像）映射到连续向量空间的技术。通过高维向量表示（如 768 维或 3072 维），模型可捕捉数据的语义信息，使得语义相似的文本在向量空间中距离更近。例如，“忘记密码”和“账号锁定”会被编码为相近的向量，从而支持语义检索而非仅关键词匹配。

### 1.1.2. <a id='toc1_1_2_'></a>[ 核心作](#toc0_)

--语义编码： \
将文本、图像等转换为向量，保留上下文信息（如 BERT 的 CLS Token 或均值池化。

--相似度计算： \
通过余弦相似度、欧氏距离（一般用余弦相似度）等度量向量关联性，支撑检索增强生成（RAG）、推荐系统等应用。

--信息降维： \
压缩复杂数据为低维稠密向量，提升存储与计算效率。

### 1.1.3. <a id='toc1_1_3_'></a>[关键技术原理](#toc0_)

上下文依赖：现代模型（如 BGE-M3）动态调整向量，捕捉多义词在不同语境中的含义。

训练方法：对比学习（如 Word2Vec 的 Skip-gram/CBOW）、预训练+微调（如 BERT）。

## 1.2. <a id='toc1_2_'></a>[主流模型分类与选型指南](#toc0_)

Embedding模型将文本转换为数值向量，捕捉语义信息，使计算机能够理解和比较内容的"意义"。

最佳实践：为特定应用测试多个embedding模型，评估在实际数据上的性能而非仅依赖通用基准。

### 1.2.1. <a id='toc1_2_1_'></a>[通用全能型](#toc0_)

BGE-M3： \
北京智源研究院开发，支持多语言、混合检索（稠密+稀疏向量），处理 8K 上下文，适合企业级知识库。

NV-Embed-v2： \
基于 Mistral-7B，检索精度高（MTEB 得分 62.65），但需较高计算资源。

### 1.2.2. <a id='toc1_2_2_'></a>[垂直领域特化型](#toc0_)

中文场景： \
BGE-large-zh-v1.5 （合同/政策文件）、M3E-base （社交媒体分析）。

多模态场景： \
BGE-VL （图文跨模态检索），联合编码 OCR 文本与图像特征。

### 1.2.3. <a id='toc1_2_3_'></a>[轻量化部署型](#toc0_)

nomic-embed-text： \
768 维向量，推理速度比 OpenAI 快 3 倍，适合边缘设备。

gte-qwen2-1.5b-instruct： \
1.5B 参数，16GB 显存即可运行，适合初创团队原型验。

### 1.2.4. <a id='toc1_2_4_'></a>[选型决策树：](#toc0_)

1. 中文为主 → BGE 系列 > M3E；
2. 多语言需求 → BGE-M3 > multilingual-e5；
3. 预算有限 → 开源模型（如 Nomic Embed）
