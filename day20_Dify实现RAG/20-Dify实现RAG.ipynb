这里的“基础 URL”设置可以参考day06_Ollama_vLLM_LMDeploy/test03.py，“基础 URL”值设置为本地服务器的地址加上:11434，和上一步配置.env中配置的保持一致。如果是Windows通过docker部署的Dify,这里的“基础 URL”设置为http://host.docker.internal:11434。

```python
#多轮对话
from openai import OpenAI

#定义多轮对话方法
def run_chat_session():
    #初始化客户端，api_key是公开的，可以随便写，但是要写
    client = OpenAI(base_url="http://localhost:8000/v1/",api_key="suibianxie")
    #初始化对话历史
    chat_history = []
    #启动对话循环
    while True:
        #获取用户输入
        user_input = input("用户：")
        if user_input.lower() == "exit":
            print("退出对话。")
            break
        #更新对话历史(添加用户输入)
        chat_history.append({"role":"user","content":user_input})
        #调用模型回答
        try:
            chat_complition = client.chat.completions.create(messages=chat_history,model="/root/autodl-tmp/model/Qwen/Qwen2.5-0.5B-Instruct")
            #获取最新回答
            model_response = chat_complition.choices[0]
            print("AI:",model_response.message.content)
            #更新对话历史（添加AI模型的回复）
            chat_history.append({"role":"assistant","content":model_response.message.content})
        except Exception as e:
            print("发生错误：",e)
            break
if __name__ == '__main__':
    run_chat_session()
```

##### 1.2.2.3.4. <a id='toc1_2_2_3_4_'></a>[保存失败处理](#toc0_)

<img src="./Image/2025-06-01-19-19-59.png" style="margin-left: 0" width="35%">

如果保存失败需要排查三个方面原因：

* 一是Windows通docker部署dify，但是ollama不是docker部署的原因。
* 二是ollama serve的IP地址，
* 三是Linux系统防火墙。

1. Windows通docker部署dify，但是ollama不是docker部署的原因。

本地windows10，用docker部署了dify，没有用docker部署ollama。

Docker进程和主机服务进行通信，这里面存在映射关系。需要改为：http://host.docker.internal:11434

host.docker.internal:
这是 Docker 提供的一个特殊主机名。当在 Docker 容器内部使用时，它会解析为主机机器上 Docker 守护进程的 IP 地址。它允许容器与主机机器上运行的服务进行通信。


2. ollama serve的IP地址排查

执行以下代码看是否可以访问

curl http://192.168.1.11:11434/api/chat -H "Content-Type: application/json" -d '{"model": "deepseek-r1:1.5b"}'

执行失败的结果：

(envollama) root@nishuixingzhou:~/AI-WSL/project/projectDify_DeepSeek/code/dify-main/docker# curl http://192.168.1.11:11434/api/chat -H "Content-Type: application/json" -d '{"model": "deepseek-r1:1.5b"}'

执行成功的结果：
(base) root@nishuixingzhou:~# curl http://192.168.1.11:11434/api/chat -H "Content-Type: application/json" -d '{"model": "deepseek-r1:1.5b"}'
{"model":"deepseek-r1:1.5b","created_at":"2025-05-31T21:42:20.692709504Z","message":{"role":"assistant","content":""},"done_reason":"load","done":true}(base) root@nishuixingzhou:~# 


执行ollama serve查看IP，要把127.0.0.1改成0.0.0.0。

为什么需要监听0.0.0.0地址？  
在计算机网络中，0.0.0.0是一个特殊的IP地址，它表示本机上的所有IPv4地址。当我们将一个服务配置为监听0.0.0.0地址时，意味着该服务将接受来自本机所有网络接口的连接，而不仅仅是某一个特定的IP地址。这使得服务更加灵活，可以在多个网络接口上同时提供服务。

(envollama) root@nishuixingzhou:~/AI-WSL/project/projectDify_DeepSeek/code/dify-main/docker# ollama serve  
Error: listen tcp 127.0.0.1:11434: bind: address already in use

[设置ollama接口能外部访问](https://blog.csdn.net/sinat_26809255/article/details/145619033?utm_medium=distribute.pc_relevant.none-task-blog-2~default~baidujs_baidulandingword~default-1-145619033-blog-144021994.235^v43^pc_blog_bottom_relevance_base3&spm=1001.2101.3001.4242.2&utm_relevant_index=3)

* 如果你使用systemd来管理服务，可以编辑Ollama的service文件：

sudo nano /etc/systemd/system/ollama.service

在[Service]部分添加以下行：

Environment="OLLAMA_HOST=0.0.0.0:11434"

保存文件并重载systemd配置

sudo systemctl daemon-reload

sudo systemctl restart ollama

* 如果你不是使用systemd，而是直接运行Ollama，可以在启动命令前设置环境变量：

export OLLAMA_HOST=0.0.0.0:11434

ollama serve

##### 1.2.2.4.1. <a id='toc1_2_2_4_1_'></a>[启动vllm API serve](#toc0_)

一、如果vllm是直接通过安装的方式部署的采用以下流程启动

例1：  
CUDA_VISIBLE_DEVICES=1 vllm serve /home/cw/llms/Qwen/Qwen1.5-1.8B-Chat --dtype=half --enforce-eager

例2：  
CUDA_VISIBLE_DEVICES=0 vllm serve /root/AI/models/Qwen/Qwen1.5-0.5B-Chat --dtype=half --enforce-eager

<img src="./Image/2025-06-01-21-57-29.png" style="margin-left: 0" width="50%">

其中
* CUDA_VISIBLE_DEVICES=0 ：在编号为0的显卡上跑；如果有多张卡，可以用该参数指定用哪张显卡跑模型。
* --dtype=half：指定模型使用半精度浮点数(FP16)进行计算和存储
* --enforce-eager：禁用CUDA的图模式，在显卡算力比较低的时候实用。

一、如果vllm是直接通过docker的方式部署的采用以下流程启动

方法1：从Docker Desktop界面运行

<img src="./Image/2025-06-02-00-10-20.png" style="margin-left: 0" width="80%">

方法2：docker run运行

先构建镜像，Dockerfile源码如下：

FROM nvcr.io/nvidia/tritonserver:25.03-vllm-python-py3
WORKDIR /app

EXPOSE 8000
CMD ["python3", "-m", "vllm.entrypoints.openai.api_server", \
    "--model", "/app/models/Qwen/Qwen2.5-1.5B-Instruct-merged", \
    "--port", "8000", "--max_model_len", "512", \
    "--tensor-parallel-size", "1", \
    "--gpu_memory_utilization", "0.85", \
    "--max_num_seqs", "128", \
    "--enforce-eager"]

然后运行

(base) root@nishuixingzhou:~# docker run -d --name vllm-server-Qwen1.5-1.8B -p 8000:8000 -v /root/AI-WSL/models/Qwen/Qwen2.5-1.5B-Instruct-merged:/app/models/Qwen/Qwen2.5-1.5B-Instruct-merged --gpus all --shm-size 16G vllm-image
